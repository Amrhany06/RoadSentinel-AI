# 06 — Spatial Hotspot Clustering (K-Means + PCA + Folium Map)
**RoadSentinel AI** — Geographic accident density clustering on real GPS coordinates.
Generates 2D PCA projection scatter plot and an interactive Folium map.

```python
import os, sys
sys.path.insert(0, '..')
import pandas as pd, joblib
from src.clustering import train_spatial_hotspots, render_hotspot_map
```

## 1. Load Geospatial Accident Coordinates

```python
data_path = '../data/us_accidents_sample.csv'
if not os.path.exists(data_path):
    from scripts.seed_data_generator import generate_all
    generate_all()

coords_df = pd.read_csv(data_path)
print(f"Loaded {len(coords_df)} geospatial accident coordinates.")
coords_df.head()
```

## 2. Fit Spatial K-Means & PCA (2D Projection)

```python
kmeans_hotspots, scaler, pca, labeled_coords, scores = train_spatial_hotspots(
    coords_df,
    k_range=range(3, 7),
    save_pca_plot='../models/metrics/spatial_pca_clusters.png'
)
joblib.dump(kmeans_hotspots, '../models/hotspot_kmeans.joblib')
print("Saved ../models/hotspot_kmeans.joblib")
```

## 3. Render Interactive Folium Hotspot Map
Precomputes static HTML map for zero-latency presentation in Streamlit.

```python
map_path = render_hotspot_map(labeled_coords, out_path='../demo/hotspot_map.html', sample_size=1500)
print(f"Interactive map saved to {map_path}")
```
