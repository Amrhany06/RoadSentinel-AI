# 01 — Exploratory Data Analysis (EDA) & Class Imbalance
**RoadSentinel AI** — Statistical exploration of incident features and class balance.
Analyzes kinetic features (speed, deceleration, bounding-box IOU overlap) and environment context.

```python
import os, sys
sys.path.insert(0, '..')
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from src.imbalance import report_class_balance
```

## 1. Load Feature Dataset
Loads engineered features (or bootstraps seed data if not yet created).

```python
data_path = '../data/engineered_features.csv'
if not os.path.exists(data_path):
    from scripts.seed_data_generator import generate_all
    generate_all()

df = pd.read_csv(data_path)
print(f"Loaded dataset shape: {df.shape}")
df.head()
```

## 2. Class Balance Analysis
Accident datasets are inherently imbalanced (safe traffic predominates). We measure the exact class ratio.

```python
balance = report_class_balance(df['is_accident'])
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='is_accident', palette=['#3B82F6', '#EF4444'])
plt.title('Accident Class Balance')
plt.xlabel('Is Accident (0 = Normal, 1 = Collision)')
plt.ylabel('Count')
plt.show()
```

## 3. Feature Distributions by Class

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.kdeplot(data=df, x='max_deceleration', hue='is_accident', fill=True, ax=axes[0], palette=['#3B82F6', '#EF4444'])
axes[0].set_title('Max Deceleration Distribution')

sns.kdeplot(data=df, x='max_iou', hue='is_accident', fill=True, ax=axes[1], palette=['#3B82F6', '#EF4444'])
axes[1].set_title('Max Box Overlap (IOU) Distribution')

sns.kdeplot(data=df, x='trajectory_variance', hue='is_accident', fill=True, ax=axes[2], palette=['#3B82F6', '#EF4444'])
axes[2].set_title('Trajectory Variance Distribution')

plt.tight_layout()
plt.show()
```

## 4. Feature Correlation Matrix

```python
num_cols = ['avg_speed', 'max_speed', 'max_deceleration', 'trajectory_variance', 'max_iou', 'is_accident']
corr = df[num_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix')
plt.show()
```
